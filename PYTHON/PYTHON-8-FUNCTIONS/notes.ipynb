```python
# pizza price calculation:
def calculate_pizza_price(size="medium"):
    """
    here am calucalting the pizza amount
    based on its size 
    and here we have types like small medium and large
    """

    base_price = 0

    if size == "small":
        base_price = 199
    elif size == "medium":
        base_price = 299
    elif size == "large":
        base_price = 399
    else:
        print("Invalid size...")
        return None

    gst = base_price * 0.18
    total_amount = base_price + gst

    return total_amount
```

```python
calculate_pizza_price()
```
