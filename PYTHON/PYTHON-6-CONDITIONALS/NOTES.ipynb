# while loop example

```python
bottle = 750 # water

while bottle > 0:
    print(f"Current water in bottle: {bottle}ml")
    bottle -= 50  # drinking 50ml of water
    print(f"Drank 50ml, remaining water: {bottle}ml")
    
print("Bottle is empty")
```

```python
meeting_started = True

cups = 0

while meeting_started:
    cups += 1
    print(f"Cups of coffee consumed: {cups}")
    if cups == 3:
        print("Meeting has started")
        meeting_started = False
```
