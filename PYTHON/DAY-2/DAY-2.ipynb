````python
print("🏏 CRICKET BATTING AVERAGE CALCULATOR")
print("="*45)

player_name = input("Enter player name: ")
total_runs = int(input("Enter total runs scored: "))
total_innings = int(input("Enter innings played: "))
not_out = int(input("Enter number of times not out: "))

# calculate
actual_innings = total_innings - not_out
batting_average = total_runs / actual_innings 

print()
print("player name:", player_name)
print("total runs:", total_runs)
print("actual innings:", actual_innings)
print("batting average:", round(batting_average, 2))
print("="*45)
print()

if batting_average >= 50:
    print("Excellent batting average!")
elif batting_average >= 40:
    print("Good batting average!")
elif batting_average >= 30:
    print("Average batting average!")
else:
    print("Below average batting average!") 

````
