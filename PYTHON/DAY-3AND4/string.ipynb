| Method | What it does | Example |
|--------|-------------|---------|
| `.upper()` | ALL CAPS | `"hello".upper()` → `"HELLO"` |
| `.lower()` | all lowercase | `"HELLO".lower()` → `"hello"` |
| `.title()` | Title Case | `"shah rukh".title()` → `"Shah Rukh"` |
| `.strip()` | Remove edge spaces | `"  hi  ".strip()` → `"hi"` |
| `.split(",")` | Break into list | `"a,b,c".split(",")` → `["a","b","c"]` |
| `",".join(lst)` | Join list to string | `",".join(["a","b"])` → `"a,b"` |
| `.replace(a,b)` | Replace text | `"Bhai".replace("B","Yaar")` |
| `.find(x)` | Find position | `"hello".find("l")` → `2` |
| `.count(x)` | Count occurrences | `"hello".count("l")` → `2` |
| `.startswith(x)` | Check start | `"Rohit".startswith("R")` → `True` |
| `.endswith(x)` | Check end | `"India".endswith("a")` → `True` |
| `x in str` | Check if exists | `"God" in "God of Cricket"` → `True` |
| `len(str)` | Length | `len("India")` → `5` |
