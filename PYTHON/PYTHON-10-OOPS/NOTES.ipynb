```python
class BankAccount:
    # contructor
    def __init__(self,ownwer,balance=0):
        self.ownwer = ownwer
        self.balance = balance



    #Methods
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited. and new balance is {self.balance}")

    def withdraw(self,amount):
        if amount > self.balance:
            print("Infusient funds")

        else:        
            self.balance-=amount
            print(f"{amount} is withrawed. and new balance is {self.balance}")
    
    def get_balance(self):
        return self.balance
```

```python
account = BankAccount('Ravikanth',1000) #(account is object)
```

```python
account.balance
```

```python
account.deposit(20000)
```

```python
account.withdraw(5000)
```
