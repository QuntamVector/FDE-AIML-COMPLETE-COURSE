```python
class BankAccount:
    # contructor
    def __init__(self,ownwer,balance=0):
        self.ownwer = ownwer
        self.balance = balance

    #Methods
    def deposit(self,amount):
        self.balance+=amount
        print(f"{amount} is deposited. and new balance is {self.balance}")

    def withdraw(self,amount):
        if amount > self.balance:
            print("Infusient funds")

        else:        
            self.balance-=amount
            print(f"{amount} is withrawed. and new balance is {self.balance}")
    
    def get_balance(self):
        return self.balance
```

# Inheritence

parent-->you will be inheriting

```python
# Parent class
class Car:
    def __init__(self,windows,doors,engine):
        self.windows = windows
        self.doors = doors
        self.engine = engine

    def drive(self):
        print(f"the person will drive {self.engine} car ")
```

```python
tata = Car(4,5,'petrol')
```

```python
tata.drive()
```

```python
#child class
class Ev(Car): # here we are inheriting the car classs
    def __init__(self,windows,doors,engine,is_selfdrive):
        super().__init__(windows,doors,engine) # calls the parent class method and also it help copy/initilise the constructor
        self.is_selfdrive = is_selfdrive

    def check_is_selfdriving(self):
        print(f"TataEv supports self driving: {self.is_selfdrive}")

```

```python
tata_ev = Ev(4,5,"electric",True)
```

```python
tata_ev.drive()
```

```python
tata_ev.check_is_selfdriving()
```

# Polymorphism

```python
# parent class

class BankAccount:
    def __init__(self,account_holder,balance) -> None:
        self.account_holder = account_holder
        self.balance = balance

    def account_details(self):
        print("-------------Account Details----------------")
        print(f"Account holder name: {self.account_holder}")
        print(f"Balance: {self.balance}")


    def calculate_intrest(self):
        print("default intrest from bankaccount")
```

```python
class SavingAccount(BankAccount):
    def calculate_intrest(self):
        intrest = self.balance * 0.04
        print(f"saving account intrest: Rs {intrest}")
```

```python
class CurrentAccount(BankAccount):
    def calculate_intrest(self):
        print(f"bank wont provide current account intrest...")
```

```python
class FixedDeposit(BankAccount):
    def calculate_intrest(self):
        intrest = self.balance * 0.075
        print(f"FD account intrest: Rs {intrest}")
```

```python
saving = SavingAccount("Manoj",20000)
```

```python
saving.calculate_intrest()
```

```python
Manju_current = CurrentAccount("Manju",30000)
```

```python
Manju_current.calculate_intrest()
```

```python
fd = FixedDeposit("Ganesh",20000)
```

```python
fd.calculate_intrest()
```
