```python
print(f"🧮 safe calculator")

while True:
    try:
        a = float(input("First number: "))
        b = float(input("second number: "))

        op = input("Operations (+,-,*,/): ")

        if op == "+":
            result = a + b
        elif op == "-":
            result = a - b
        elif op == "*":
            result = a * b
        elif op == "/":
            result = a / b
        else:
            print("❌ invalid  operator USE ONLY OPERATION (+,-,*,/) ")   

        print(f"✅ Result: {a} {op} {b} = {result}")

    except ValueError:
        print("❌ please enter number only...")
    except ZeroDivisionError:
        print("❌ you cannot divide by zero...")
    except Exception as e:
        print("❌ unexcpected error : {e}")

    finally:
        print("Calculation attempt is completed")

    again = input("Do you want to calculate again (yes/no): ").lower()
    if again != "yes":
        print("👋🏻 Good Bye...")
        break

    
```

```python
def calculator():
    print(f"🧮 safe calculator")

    while True:
        try:
            a = float(input("First number: "))
            b = float(input("second number: "))

            op = input("Operations (+,-,*,/): ")

            if op == "+":
                result = a + b
            elif op == "-":
                result = a - b
            elif op == "*":
                result = a * b
            elif op == "/":
                result = a / b
            else:
                print("❌ invalid  operator USE ONLY OPERATION (+,-,*,/) ")   

            print(f"✅ Result: {a} {op} {b} = {result}")

        except ValueError:
            print("❌ please enter number only...")
        except ZeroDivisionError:
            print("❌ you cannot divide by zero...")
        except Exception as e:
            print("❌ unexcpected error : {e}")

        finally:
            print("Calculation attempt is completed")

        again = input("Do you want to calculate again (yes/no): ").lower()
        if again != "yes":
            print("👋🏻 Good Bye...")
            break

    
```
