# Quantam Vector Restaurant Management System

menu = {
"chicken_biryani": 220,
"veg_biryani": 200,
"cold_coffee": 100,
"garlic_nan": 150
}

order = []          # Stores ordered items
customers = set()   # Stores unique customers
sales = {}          # Stores item-wise sales count

print("\n--- Welcome to Quantam Vector Restaurant ---")

customer = input("Enter customer name: ").strip().title()

if not customer:
customer = "Guest"

customers.add(customer)

# ---------------- MENU ----------------

print("\n--- MENU ---")

for item, price in menu.items():
    display_name = item.replace("_", " ").title()
    print(f"{display_name:<25} Rs {price}")

print("\nType 'done' to finish your order.")

# ---------------- ORDER ----------------

while True:

```
item = input("\nEnter item: ").strip().lower().replace(" ", "_")

# Finish order
if item == "done":
    break

# Check whether item exists
if item in menu:

    order.append(item)

    # Count sales
    if item in sales:
        sales[item] += 1
    else:
        sales[item] = 1

    print(f"{item.replace('_', ' ').title()} added to your order.")

else:
    print("Sorry! Item is not available on the menu.")
    print("Please choose an item from the menu.")
```

# ---------------- BILL ----------------

print("\n========== BILL ==========")
print(f"Customer: {customer}")

if not order:

```
print("No items ordered.")
```

else:

```
total = 0

for item in order:

    price = menu[item]
    display_name = item.replace("_", " ").title()

    print(f"{display_name:<25} Rs {price}")

    total += price

print("--------------------------")
print(f"Total Bill:               Rs {total}")
```

# ---------------- SUMMARY ----------------

print("\n========== SUMMARY ==========")

print(f"Customer: {customer}")
print(f"Items Ordered: {len(order)}")

print("\nOrder List:")

for item in order:
print(f"- {item.replace('_', ' ').title()}")

print(f"\nUnique Customers: {customers}")

print(f"\nSales Dictionary: {sales}")

# ---------------- BEST SELLING ITEM ----------------

if sales:

```
best_item = max(sales, key=sales.get)

print(
    f"\nBest Selling Item: "
    f"{best_item.replace('_', ' ').title()} "
    f"({sales[best_item]} sold)"
)
```

else:

```
print("\nNo sales recorded.")
```

print("\nThank you for visiting Quantam Vector Restaurant! 😊")
