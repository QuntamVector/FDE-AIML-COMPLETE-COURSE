## PyPDF (pypdf)

- Pure Python library.
- Extracts text and accesses PDF pages and metadata.
- Good for straightforward text-based PDFs.
- Often a simple choice for basic LangChain RAG projects.

## PyMuPDF (pymupdf)

- Python bindings for the MuPDF engine.
- Often faster for large or complex PDFs.
- Supports text extraction, page rendering, images, and layout-related information.
- Useful when processing PDFs with tables, images, or more complex layouts, though extraction quality depends on the document.

```python
from langchain_community.document_loaders import PyMuPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_core.documents import Document
from typing import List


class SmartPdfProcessor:

    """
    Advance preprocesing with error handling...
    """

    def __init__(self,chunk_size=1000,chunk_overlap=200):
        self.chunk_size = chunk_size
        self.chunk_overlap = chunk_overlap
        self.text_splitter = RecursiveCharacterTextSplitter(
            chunk_size = chunk_size,
            chunk_overlap = chunk_overlap,
            separators= [" "]
        )

    def _clean_text(self,text:str) -> str:
        """ 
        Cleaning the text by removing special characters numbers and convert into lowercase
        """
        text = " ".join(text.split())

        # fix ligerature
        text = text.replace("ﬁ", "fi")
        text = text.replace("ﬂ", "fl")

        return text


    def process_pdf(self,pdf_path:str) -> List[Document]:
        """ 
        Processing the pdf with smart chunking and metadata enrichment
        """
        loader=PyMuPDFLoader(pdf_path)
        pages = loader.load()


        for page_num,page in enumerate(pages):
            """ 
            Cleanig the text
            """

            processed_chunks = []

            cleaned_text = self._clean_text(page.page_content)

            ## if i have empty pages skip those
            if len(cleaned_text.strip()) < 50:
                continue

            # create a text with enhanced metadata enrichment

            chunks = self.text_splitter.create_documents(
                texts = [cleaned_text],
                metadatas=[{
                    **page.metadata,
                    "page": page_num+1,
                    "total_pages":len(pages),
                    "chunk_method": "SmartPdfProcessor",
                    "char_count": len(cleaned_text),
                }]
            )

            processed_chunks.extend(chunks)

        return processed_chunks
        
```

```python
preprocer = SmartPdfProcessor()
```

```python
smartchunk = preprocer.process_pdf("pdfdoc/fp_monitoring_2000_frep_09_en.pdf")
```
